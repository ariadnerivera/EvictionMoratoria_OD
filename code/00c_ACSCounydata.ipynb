```python
data2.rename(columns = {"Geo_FIPS": "geofips",
                        "SE_A00001_001": "totpop", #total population
                        "SE_A14006_001": "mhhinc", #median household income
                        "SE_A10032_001":"housingunits", #number of housing units
                        "SE_A18002_001": "renterhousingunits", #numbre of renter housing units"
                        "SE_A10062B_001": "poprenterhousing", #population in household renter units
                        "SE_B18002_002": "rent30_49inc", 
                        "SE_B18002_003": "rent50plusinc", 
                        "SE_A13003B_001": "poppovertystatus", 
                        "SE_A13003B_002": "pop18_64poverty",
                        "SE_A14028_001": "gini"}, inplace = True)

data2["Year"] = year
```

```python
for year in range(2017, 2021):
    filename = "/1_raw/census_acs/acs_"+str(year)+".csv"
    data = pd.read_csv(fp+filename)
    
    data2 = data[["Geo_FIPS", "Geo_NAME", "SE_A00001_001", "SE_A14006_001", "SE_A10032_001", "SE_A18002_001", "SE_A10062B_001", 
              "SE_B18002_002", "SE_B18002_003", "SE_A13003B_001", "SE_A13003B_002", "SE_A14028_001"]]
    
    data2.rename(columns = {"Geo_FIPS": "geofips",
                        "SE_A00001_001": "totpop",
                        "SE_A14006_001": "mhhinc", 
                        "SE_A10032_001":"housingunits",
                        "SE_A18002_001": "renterhousingunits",
                        "SE_A10062B_001": "poprenterhousing", 
                        "SE_B18002_002": "rent30_49inc", 
                        "SE_B18002_003": "rent50plusinc", 
                        "SE_A13003B_001": "poppovertystatus", 
                        "SE_A13003B_002": "pop18_64poverty",
                        "SE_A14028_001": "gini"}, inplace = True)
    
    data2["Year"] = year
    

    # Write to a csv file
    data2.to_csv(fp+"/2_intermediate/census_acs/"+"ACSclean_"+str(year)+".csv")

    print("Done!")
```

```python
all_files = glob.glob(os.path.join(fp+"/2_intermediate/census_acs/", "*.csv"))     # advisable to use os.path.join as this makes concatenation OS independent
df_from_each_file = (pd.read_csv(f) for f in all_files)
concatenated_acs   = pd.concat(df_from_each_file, ignore_index=True)
concatenated_acs.head()
```
